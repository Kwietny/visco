```python
import pandas as pd
from openpyxl import Workbook
from openpyxl.styles import PatternFill
from openpyxl.utils.dataframe import dataframe_to_rows
```

```python
visco = pd.read_csv('../data/viscoelastic_cleaned.csv', keep_default_na=False, na_values='')
```

```python
visco
```

```python
drop_columns = [
    'id_surfactant1',
    'smiles_surfactant1',
    'id_surfactant2',
    'smiles_surfactant2',
    'id_additive1',
    'smiles_additive1',
    'id_additive2',
    'smiles_additive2',
    'type_surfactant1',
    'type_surfactant2'
]
X = visco.drop(drop_columns, axis=1)
```

```python
X
```

```python
label_check_window = visco.dropna(subset=['eta_0', 'viscoelasticity']).sort_values(by='eta_0').copy()

label_check_window['neighbor_sum'] = label_check_window['viscoelasticity'].rolling(window=5, center=True, min_periods=1).sum()

possible_typos = label_check_window[
    (label_check_window['viscoelasticity'] == 0) &
    (label_check_window['neighbor_sum'] >= 4)
]
```

```python
output_path = '../data/viscoelastic_marked_sorted.xlsx'

visco_sorted = visco.copy()

visco_sorted.insert(0, 'Original_Excel_Row', visco_sorted.index + 2)

visco_sorted = visco_sorted.sort_values(by='eta_0')

visco_sorted = visco_sorted.reset_index(drop=True)

typo_new_positions = visco_sorted[
    visco_sorted['Original_Excel_Row'].isin(possible_typos.index + 2)
].index

wb = Workbook()
ws = wb.active
ws.title = "Sorted Viscoelastic Data"

for r in dataframe_to_rows(visco_sorted, index=False, header=True):
    ws.append(r)

yellow_fill = PatternFill(start_color='FFFF00', end_color='FFFF00', fill_type='solid')

for row_idx in typo_new_positions:
    excel_row = row_idx + 2
    for cell in ws[excel_row]:
        cell.fill = yellow_fill

wb.save(output_path)
```
