```python
from skfp.fingerprints import BCUT2DFingerprint
from modules.data_loader import *
```

```python
visco = load_cleaned()
```

```python
smiles = visco['smiles_surfactant1']
smiles = pd.concat([smiles, visco['smiles_surfactant2']])
smiles = pd.concat([smiles, visco['smiles_additive1']])
smiles = pd.concat([smiles, visco['smiles_additive2']])
smiles = smiles.drop_duplicates()
smiles = smiles.dropna()
len(smiles)
```

```python
fp = BCUT2DFingerprint(partial_charge_model='formal')
```

```python
features = fp.transform(smiles. values)
```

```python
smiles = smiles.to_frame(name='smiles')

feature_names = [
    'burden_eigenvalue_mass_max',
    'burden_eigenvalue_mass_min',
    'burden_eigenvalue_charge_max',
    'burden_eigenvalue_charge_min',
    'burden_eigenvalue_logp_max',
    'burden_eigenvalue_logp_min',
    'burden_eigenvalue_mr_max',
    'burden_eigenvalue_mr_min'
]

bcut= pd.DataFrame(
    features,
    columns=feature_names,
    index=smiles.index
)

bcut2d = pd.concat([smiles, bcut], axis=1)
```
